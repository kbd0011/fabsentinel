# 03 — Multivariate SPC: PCA + Hotelling T² + SPE

```python
import pandas as pd, matplotlib.pyplot as plt
mstats = pd.read_csv('artifacts/multivariate_stats.csv', index_col=0)
fig, ax = plt.subplots(figsize=(7, 5))
colors = ['#c53030' if x else '#4299e1' for x in mstats['t2_ooc']]
ax.scatter(mstats['spe'], mstats['t2'], c=colors, alpha=0.6, s=10)
ax.set_xlabel('SPE residual'); ax.set_ylabel('Hotelling T²'); ax.set_title('Multivariate SPC')
plt.show()
```

```python
# Contribution plot for the worst-T² lot
import joblib
from fabsentinel.spc.multivariate import contribution_plot
from fabsentinel.vm.features import prepare
df = pd.read_parquet('data/cleaned.parquet')
sigcols = [c for c in df.columns if c.startswith('signal_')]
X, _ = prepare(df, sigcols)
mspc = joblib.load('artifacts/mspc_model.joblib')
worst = int(mstats['t2'].idxmax())
contrib = contribution_plot(mspc, X.iloc[worst], top_k=10)
contrib.plot.barh(figsize=(6, 5)); plt.gca().invert_yaxis(); plt.title(f'T² contributions — lot {worst}'); plt.show()
```
