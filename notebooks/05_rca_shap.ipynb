```python
import pandas as pd
shap_global = pd.read_csv('artifacts/shap_global.csv', index_col=0, header=None, names=['mean_abs_shap'])
shap_global.head(15).plot.barh(figsize=(7, 8))
import matplotlib.pyplot as plt
plt.gca().invert_yaxis(); plt.title('Top-15 SHAP-attributed signals (global)'); plt.show()
```

```python
# Local attribution for the worst-T² lot
from fabsentinel.vm.train import load
from fabsentinel.vm.features import prepare
from fabsentinel.rca.shap_attribution import fit, top_features_for_lot
artifact = load('artifacts/vm_model.joblib')
df = pd.read_parquet('data/cleaned.parquet')
sigcols = [c for c in df.columns if c.startswith('signal_')]
X, _ = prepare(df, sigcols)
mstats = pd.read_csv('artifacts/multivariate_stats.csv', index_col=0)
worst = int(mstats['t2'].idxmax())
report = fit(artifact, X, n_background=50)
local = top_features_for_lot(report, X.iloc[worst], top_k=10)
local
```
