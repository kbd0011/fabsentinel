# 01 — Data quality profile

Missingness and low-variance signals. SECOM is messy by design.

```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_parquet('data/processed.parquet')
signal_cols = [c for c in df.columns if c.startswith('signal_')]
missing = df[signal_cols].isna().mean().sort_values(ascending=False)
print(f"Signals with >50% missing: {(missing > 0.5).sum()}")
missing.head(40).plot.barh(figsize=(7, 8)); plt.gca().invert_yaxis(); plt.title('Top-40 highest-missing signals'); plt.show()
```

```python
var = df[signal_cols].var().sort_values()
print(f"Signals with ~zero variance: {(var < 1e-6).sum()}")
```
