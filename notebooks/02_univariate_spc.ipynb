# 02 — Univariate SPC

Shewhart, EWMA, CUSUM, plus run-rule overlays.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from fabsentinel.spc.shewhart import fit_individuals, detect
from fabsentinel.spc.ewma import fit_and_detect as ewma_run
from fabsentinel.spc.cusum import fit_and_detect as cusum_run

df = pd.read_parquet('data/cleaned.parquet')
sigcols = [c for c in df.columns if c.startswith('signal_')]
top = df[sigcols].var().sort_values(ascending=False).head(5).index.tolist()
print('Top-5 highest-variance signals:', top)
```

```python
sig = top[0]
arr = df[sig].values
sh = fit_individuals(arr)
fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(arr, color='steelblue', lw=0.8)
ax.axhline(sh.mean, color='green', ls='--', lw=0.7)
ax.axhline(sh.ucl, color='red', ls='--', lw=0.7)
ax.axhline(sh.lcl, color='red', ls='--', lw=0.7)
ax.set_title(f'{sig} — Shewhart individuals'); plt.show()
print(f'mean={sh.mean:.3f}, sigma={sh.sigma:.3f}, OOC = {int(detect(arr, sh).sum())}')
```
