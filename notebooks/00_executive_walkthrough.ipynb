A 5-cell tour of the full FabSentinel pipeline on SECOM. Run `make demo` from the repo root first so the artifacts exist.

```python
import pandas as pd
import json
from pathlib import Path

# Quality report
q = json.loads(Path('artifacts/quality_report.json').read_text())
print('SECOM after ingest + quality:')
print(f"  lots = {q['n_lots']}, signals = {q['n_signals']}, fails = {q['n_fails']}")
print(f"  pass:fail ratio = {q['pass_fail_ratio']:.1f}:1")
print(f"  dropped (missing) = {len(q['dropped_missing'])}, dropped (low-var) = {len(q['dropped_lowvar'])}, retained = {q['retained_count']}")
```

```python
# Multivariate SPC
mstats = pd.read_csv('artifacts/multivariate_stats.csv', index_col=0)
print(f"T² OOC: {mstats['t2_ooc'].sum()} of {len(mstats)} lots")
print(f"SPE OOC: {mstats['spe_ooc'].sum()} of {len(mstats)} lots")
print('Top-5 T² lots:')
display(mstats.nlargest(5, 't2'))
```

```python
# SHAP global ranking
shap_top = pd.read_csv('artifacts/shap_global.csv', index_col=0, header=None, names=['mean_abs_shap'])
shap_top.head(10)
```
